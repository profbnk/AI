# 🌊 Interactive Sine Wave Animations with Jupyter Widgets

This notebook demonstrates how to build interactive mathematical visualizations directly inside a Google Colab notebook using **ipywidgets** and **matplotlib**.

---

## 🛠️ Concepts & Components Covered

### 1. Basic Interactive Controls (`ipywidgets`)
We started by exploring basic interactive controls like `widgets.IntSlider` to read user inputs dynamically.

### 2. Live Dynamic Parameter Tuning (`widgets.interact`)
Using `widgets.interact`, we bound variables directly to functions. This allows real-time rendering of a sine wave $y = \sin(x + t)$ as the shift parameter $t$ is scrubbed.

### 3. Automatic Animation (`widgets.Play`)
Instead of dragging a slider manually, we integrated the `widgets.Play` widget. This steps through a sequence of frames automatically to animate the phase and frequency changes:
- **Phase Modulation:** $y = \sin(x + 0.1 \cdot t)$
- **Frequency & Phase Controls:** $y = \sin(\text{freq} \cdot (x + 0.1 \cdot t))$

---

## 🚀 Features
- **Real-Time Physics/Wave Simulation:** Visualize wave mechanics instantly.
- **Fluid Controls:** Play, pause, step, and reset your animation cleanly.
- **Multi-parameter Modulation:** Adapt frequency and phase on-the-fly.

```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
```

```python
def sine_wave(t=0):
  y = np.sin(x + t)
  plt.plot(x,y)
```

```python
def sine_wave_slow(t=0):
  y = np.sin(x + 0.1*t)
  plt.plot(x,y)
```

```python
sine_wave(t=2)
```

```python
widgets.interact(sine_wave, t=(0, 20, 1))
```

```python
widgets.interact(sine_wave, t=widgets.Play(min=0, max=20))
```

```python
def sine_wave_with_freq(t=0, freq=1):
  y = np.sin(freq*(x + 0.1*t))
  plt.plot(x,y)
```

```python
widgets.interact(sine_wave_with_freq, t=widgets.Play(min=0, max=20), freq=(1, 5, 0.5))
```
